The first column contains the market factor values, i.e., the monthly excess returns on the market portfolio over the risk-free rate, and the second column contains the monthly risk-free rates. All these numbers are expressed as percentage values, which means that a value of 0.12 in the column RF means a risk-free rate of 0.12%.

We note that the two DataFrames have different indexes because the data from heterogenous sources are formatted differently. This can create problems later when we want to match the data from one file with the data from the other, as Python will check the indexes. To avoid these problems, we are converting every index in the DataFrame of values to a number that represents the date in the format yyyymm. To do this, we call the function `strftime`, which converts a date (technically, a TimeStamp object) into a string with a specified format, and we convert the string into an integer with `int`. The DataFrame method `rename` changes the index values by applying a function to them.

```python
val = val_raw.rename(index=lambda x: int(x.strftime('%Y%m')))
```

```python
ret = val.pct_change().iloc[1:]
mkt = ffr['Mkt-RF'] / 1e2
rf = ffr['RF'] / 1e2
```

Let us calculate the monthly returns on the fund and the stock, and then subtract the risk-free rate to obtain the excess returns. To subtract the column of risk-free rates from every column of returns, we use the DataFrame method `sub` with the argument `axis=0` to tell Pandas to match the indexes (axis 0) of the two DataFrames.

```python
xret = ret.sub(rf, axis=0)
xret.head()
```
